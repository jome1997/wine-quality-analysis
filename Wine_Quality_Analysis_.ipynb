```python
import pandas as pd
wine = pd.read_csv("winequality-red.csv")
print(wine.head())
print(wine.shape)
```

```python
print(wine.columns.tolist())
print(wine.isna().sum())
print(wine.duplicated().sum())
```

```python
print(wine.describe())
```

```python
print(wine.corr()["quality"].sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt

wine.groupby("quality")["alcohol"].mean().plot(kind="bar")
plt.title("Average Alcohol by Quality")
plt.show()
```

```python
wine.groupby("quality")["volatile acidity"].mean().plot(kind="bar")
plt.title("Average Volatile Acidity by Quality")
plt.show()
```

```python
print(wine["quality"].value_counts().sort_index())
```
